# Pandas
Pandas is a Python library used for working with data sets.

It has functions for analyzing, cleaning, exploring, and manipulating data.

### Keywords
- **pandas**: An open-source Python library used for analyzing, cleaning, exploring, and manipulating data. It is one of the most used tools in data science.
- **%pip install**: A Jupyter magic command that installs a Python package into the environment the notebook is running in.
- **import / alias (`pd`)**: `import` loads a library into your code. Writing `import pandas as pd` gives pandas the short alias `pd`, which is the standard convention.
- **`__version__`**: An attribute that tells you which version of a library is installed. It is useful for checking compatibility.
- **DataFrame**: A 2-dimensional table of rows and columns in pandas. It can be created from a Python dictionary, where each key becomes a column.

## Why Use Pandas?
Pandas allows us to analyze big data and make conclusions based on statistical theories.

Pandas can clean messy data sets, and make them readable and relevant.

Relevant data is very important in data science.

## Creating a dataframe

```python
mydataset = {  # Create a Python dictionary that will hold our data
  'cars': ["BMW", "Volvo", "Ford"],  # Key 'cars' becomes a column with three car brands
  'passings': [3, 7, 2]  # Key 'passings' becomes a column with three numbers
}  # End of the dictionary

myvar = pd.DataFrame(mydataset)  # Convert the dictionary into a pandas DataFrame (a table)

print(myvar)  # Display the DataFrame
```
