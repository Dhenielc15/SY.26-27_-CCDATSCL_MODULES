# Pandas DataFrame

A Pandas DataFrame is a 2 dimensional data structure, like a 2 dimensional array, or a table with rows and columns.

### Keywords
- **DataFrame**: A 2-dimensional data structure in pandas, like a table with rows and columns. Each column of a DataFrame is a Series.
- **`pd.DataFrame()`**: The function used to create a DataFrame, for example from a dictionary of lists.
- **`.loc`**: An attribute used to select rows (and columns) by their label. A single label returns a Series, while a list of labels returns a DataFrame.
- **Named index**: Custom row labels (e.g., "day1", "day2") given with the `index` argument instead of the default 0, 1, 2.

```python
import pandas as pd  # Import pandas with the standard alias 'pd'

data = {  # Create a dictionary where each key will become a column
  "calories": [420, 380, 390],  # Column 'calories' with three values
  "duration": [50, 40, 45]  # Column 'duration' with three values
}  # End of the dictionary

#load data into a DataFrame object:
df = pd.DataFrame(data)  # Convert the dictionary into a DataFrame

print(df)  # Display the DataFrame
```

## Locate Row
Pandas use the loc attribute to return one or more specified row(s)

This will return a pandas series

```python
print(df.loc[0])  # Use .loc to get the row with index 0 (returned as a Series)
```

#### Return row 0 and 1:

Note: When using **[]**, the result is a **Pandas DataFrame**.

```python
print(df.loc[[0, 1]])  # Pass a list of indexes to .loc to get rows 0 and 1 (returned as a DataFrame)
```

## Named indexes

#### Add a list of names to give each row a name:

```python
data = {  # Create the same dictionary of data again
  "calories": [420, 380, 390],  # Column 'calories' with three values
  "duration": [50, 40, 45]  # Column 'duration' with three values
}  # End of the dictionary

df = pd.DataFrame(data, index = ["day1", "day2", "day3"])  # Create a DataFrame with named row indexes

print(df)  # Display the DataFrame with the named indexes
```

#### Locate named indexes

Use the named index in the loc attribute to return the specified row(s).

```python
print(df.loc["day2"])  # Use .loc with a named index to get the row labeled "day2"
```
