### Keywords
- **Series**: A one-dimensional labeled array that can hold any data type. Each column in a DataFrame is a Series.
- **DataFrame**: A 2-dimensional table of rows and columns, similar to a spreadsheet or a SQL table.
- **dtype / `.dtypes`**: The data type of a Series or column (e.g., `object`, `int64`, `float64`, `datetime64`). `.dtypes` shows the type of every column.
- **`merge()` / Joins**: Combines two DataFrames based on matching values. The `how` parameter sets the join type: `inner`, `outer`, `left`, `right`, or `cross`.
- **kagglehub**: A library for downloading datasets directly from Kaggle using `dataset_download('username/dataset-name')`.
- **`pd.read_csv()`**: Reads a CSV file into a DataFrame.
- **`.head()` / `.tail()`**: Show the first or last 5 rows of a DataFrame for a quick preview.
- **`.describe()`**: Gives summary statistics (count, mean, std, min, quartiles, max) for numeric columns.
- **`.loc` / `.set_index()`**: `.loc` selects rows by their index label. `.set_index()` turns an existing column into the row index.
- **`.shape` / `.columns`**: `.shape` gives the number of rows and columns, while `.columns` lists the column names.
- **Subsetting**: Selecting only the rows or columns you need. Rows can be filtered with a condition inside `.loc[]` or with `.query()`.
- **Missing values (`.isna()`, `~`)**: `.isna()` returns `True` where a value is missing (NaN). The tilde `~` reverses the condition to keep rows that are NOT missing.
- **Casting (`.astype()`, `pd.to_datetime()`, `pd.to_numeric()`)**: Converting a column from one data type to another, such as text to number or text to date.
- **New column**: Created by assigning a calculation to a new column name, like `df['ratio'] = df['a'] / df['b']`.
- **`pd.concat()`**: Stacks DataFrames together, which can be used to append rows.
- **`.plot()`**: Creates charts (histogram, scatter, etc.) directly from a DataFrame using matplotlib.
- **`.to_csv()`**: Saves a DataFrame to a CSV file.

## Series

A Pandas Series is a one-dimensional labeled array capable of holding data of various types, including integers, strings, floating-point numbers, and Python objects.

We can manually create a Series from scratch. To do this, we initially define a Python `list`

```python
mydata = ['Boat', 'Car', 'Bike', 'Truck']  # Create a Python list of four strings
```

And pass the python `list` as an argument to `pd.Series()`

```python
myseries1 = pd.Series(mydata)  # Convert the list into a pandas Series
```

If we print our series..

```python
print(myseries1)  # Display the Series with its index and dtype (object = text)
```

We can see that each value will have an `index` and also the datatype of the series. In this example, the values are all `object` which is similar to a `string` type in Python

If we create another series with integers,

```python
mydata = [1, 55, 99, 43]  # Create a Python list of four integers
myseries2 = pd.Series(mydata)  # Convert the integer list into a Series
print(myseries2)  # Display the Series; dtype will be int64
```

Our new series will have a dtype of `int64` which means myseries2 contains integers

## DataFrame

A DataFrame is a 2-dimensional data structure that can store data of different types (including characters, integers, floating point values, categorical data and more) in columns. It is similar to a spreadsheet, a SQL table

We can merge two separate dataframes togethet by using the `merge()` method.

For example if we have two dataframes

```python
data1 = {  # Create a dictionary for the first DataFrame
  "name": ["Sally", "Mary", "John", "ABC"],  # Column 'name' with four names
  "age": [50, 40, 30, 99]  # Column 'age' with four ages
}  # End of the dictionary

df1 = pd.DataFrame(data1)  # Convert the dictionary into the first DataFrame
```

```python
data2 = {  # Create a dictionary for the second DataFrame
  "name": ["Sally", "Peter", "Micky", "ABC"],  # Column 'name' with four names (some overlap with df1)
  "age": [77, 44, 22, 99]  # Column 'age' with four ages
}  # End of the dictionary

df2 = pd.DataFrame(data2)  # Convert the dictionary into the second DataFrame
```

There are four basic ways to handle the join (inner, left, right and outer)

<img src="https://media.geeksforgeeks.org/wp-content/uploads/joinimages.png"/>

```python
df1.head()  # Display the first rows of df1
```

```python
df2.head()  # Display the first rows of df2
```

```python
merge_df = df1.merge(df2, how='inner')  # Inner join: keep only rows that match in BOTH DataFrames
merge_df  # Display the merged result
```

```python
merge_df = df1.merge(df2, how='outer')  # Outer join: keep ALL rows from both DataFrames
merge_df  # Display the merged result
```

```python
merge_df = df1.merge(df2, how='left')  # Left join: keep all rows from df1 and matching rows from df2
merge_df  # Display the merged result
```

```python
merge_df = df1.merge(df2, how='right')  # Right join: keep all rows from df2 and matching rows from df1
merge_df  # Display the merged result
```

```python
merge_df = df1.merge(df2, how='cross')  # Cross join: pair EVERY row of df1 with EVERY row of df2
merge_df  # Display the merged result
```

# Reading in Data

Pandas provides a suite of read functions designed to import data from various file formats and data sources into a DataFrame. These functions are crucial for data loading and preprocessing in Python.

There are multiple ways to retrieve file/dataset. In this example, we are downloading the file directly from kagglehub by using the `dataset_download` function.

First, we need to import the kagglehub package. In kagglehub, datasets are commonly found in other users accounts. To retrieve a dataset from an account in kaggle, we need to define the username and the name of the dataset.


```python
import kagglehub  # Import the kagglehub library for downloading Kaggle datasets

filepath = kagglehub.dataset_download('robikscube/mr-beast-youtube-video-statistics')  # Download the dataset ('username/dataset-name') and get its folder path
```

The `dataset_download` function returns a file path or a directory of the dataset and not the file itself. To verify that it is a path, we can use the `os.path.isdir()` function

```python
import os  # Import the os module for working with files and folders

if os.path.isdir(filepath):  # Check if the downloaded path is a directory (folder)
  print(True)  # Print True if it is a directory
```

Then use `os.listdir()` to check the files inside the path or directory.

```python
contents = os.listdir(filepath)  # Get a list of all files inside the downloaded folder
contents  # Display the list of files
```

```python
mydataset = filepath + "/" + contents[0]  # Build the full path to the first file in the folder
mydataset  # Display the full file path
```

We can use `pd.read_csv` to read a comma-separated values (csv) file into `DataFrame`.

```python
df = pd.read_csv(mydataset)  # Read the CSV file into a DataFrame
```

One very important aspect of inspecting our data is viewing the statistical summary of our dataframe.


When dealing with a massive dataset, it is very helpful to know some insights of what we are dealing with.

The `.describe()` is a quick way to get a statistical summary of our data in a DataFrame and will show answer key questions that include

- What’s the average value?
- Are there outliers?
- How spread out is the data?
- What are the trends and anomalies?





```python
df.describe()  # Show summary statistics (count, mean, std, min, quartiles, max) of numeric columns
```

Here’s what each part means:

- `count`: Number of non-null (non-missing) entries.
- `mean`: The average value.

- `std` (Standard Deviation): Tells you how spread out the data is. A high value means more variability.

- `min`: The smallest value in the column.

25%, 50%, 75% (Percentiles/Quartiles):

- `25%`: The first quartile — 25% of the data falls below this value.

- `50%`: The median (middle value).

- `75%`: The third quartile — 75% of the data falls below this value.

- `max`: The highest value in the column.

## Subsetting rows

Let us say that we want to filter rows that has a view count of more than 1 million. There are two ways to filter rows.

We can use `.loc`

```python

df_subset1 = df.loc[df['viewCount'] > 1_000_000]  # Filter with .loc: keep rows where viewCount is greater than 1 million
df_subset1  # Display the filtered rows
```

Or we can use `.query`

```python
df_subset2 = df.query('viewCount > 1000000')  # Filter with .query(): same condition written as a string
df_subset2  # Display the filtered rows
```

## Subsetting undefined or missing values

Subsetting is very useful for spotting missing values in a `DataFrame`. We can use the `.na()` function to create a subset of columns with missing values.

```python
df['viewCount'].isna()  # Return True for each row where 'viewCount' is missing (NaN), otherwise False
```

Running this will display a boolean value for each row in the viewCount column. Notice the `True` value for the third row indicating that it has a missing value in this column.

We can also use `.na()` to create a subset of rows with missing values.

```python
df.loc[df['viewCount'].isna()]  # Show only the rows where 'viewCount' is missing
```

Using tilde `~` acts as a negation where it will instead display the rows that does not have a missing value in the viewCount column

```python
df = df.loc[~df['viewCount'].isna()]  # Use ~ (NOT) to keep only rows where 'viewCount' is NOT missing
```

```python
df = df.loc[~df['likeCount'].isna()]  # Keep only rows where 'likeCount' is NOT missing
```

# Casting dtypes

The `astype()` method in pandas is used to cast a pandas object to a specified data type.


Common use cases
- Converting string representations of numbers to actual numeric types (integers or floats) to enable mathematical operations.
- Converting numeric types to strings for display or specific data handling requirements.
- Converting object columns containing dates to datetime objects for time-series analysis (often used in conjunction with `pd.to_datetime()`).

Let us say that we want to convert `viewCount` and `likeCount` to an `integer`

```python
df.dtypes  # Check the current data types before casting
```

To do this, we can use the `.astype()` function.

```python
df['viewCount'] = df['viewCount'].astype('int')  # Convert 'viewCount' to integer type
df['likeCount'] = df['likeCount'].astype('int')  # Convert 'likeCount' to integer type
```

```python
df.dtypes  # Check the data types again to confirm the change
```

There is another column that we need to cast. Notice that publishTime is an object type.

If we look closely, it contains year, month, date and time which should be in  datetime format

```python
df['publishTime']  # Display the 'publishTime' column (currently stored as text/object)
```

We can use the `.to_datetime()` function to cast it to a datetime type.

```python
df['publishTime'] = pd.to_datetime(df['publishTime'])  # Convert 'publishTime' from text to datetime type
```

We can also cast integers to strings

```python
df['likeCount'] = pd.to_numeric(df['likeCount'].astype('str'))  # Cast 'likeCount' to string, then convert it back to a number with pd.to_numeric()
```
