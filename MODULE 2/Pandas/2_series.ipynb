# Pandas Series

A Pandas Series is like a column in a table.

It is a one-dimensional array holding data of any type.

### Keywords
- **Series**: A one-dimensional labeled array in pandas that can hold any data type. Think of it as a single column in a table.
- **`pd.Series()`**: The function used to create a Series, for example from a Python list.
- **Index / Labels**: The label attached to each value in a Series. By default the labels are numbers starting at 0.
- **`index` argument**: A parameter of `pd.Series()` that lets you give your own custom labels (e.g., "x", "y", "z").
- **Accessing values**: You can get a value from a Series by putting its label inside square brackets, like `myvar["x"]`.

```python
import pandas as pd  # Import pandas with the standard alias 'pd'

a = [1, 7, 2]  # Create a regular Python list of three integers

myvar = pd.Series(a)  # Convert the list into a pandas Series (a labeled 1D array)

print(myvar)  # Display the Series together with its default index (0, 1, 2)
```

## Labels

If nothing else is specified, the values are labeled with their index number. First value has index 0, second value has index 1 etc.

This label can be used to access a specified value.

```python
print(myvar[0])  # Access and print the value at index label 0
print(myvar[1])  # Access and print the value at index label 1
print(myvar[2])  # Access and print the value at index label 2
```

## Create labels

#### With the index argument, you can name your own labels.

```python
myvar = pd.Series(a, index = ["x", "y", "z"])  # Create a Series with custom labels x, y, z instead of 0, 1, 2

print(myvar)  # Display the Series with its custom labels
```

#### When you have created labels, you can access an item by referring to the label.

```python
print(myvar["x"])  # Access and print the value labeled "x"
print(myvar["y"])  # Access and print the value labeled "y"
print(myvar["z"])  # Access and print the value labeled "z"
```
