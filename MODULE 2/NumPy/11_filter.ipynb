# Filter Numpy Array

Getting some elements out of an existing array and creating a new array out of them is called filtering.

#### Create an array from the elements on index 0 and 2:

```python
import numpy as np

# Step 1: Create an array with four numbers
arr = np.array([41, 42, 43, 44])

# Step 2: Create a boolean filter list (True means "keep", False means "drop")
x = [True, False, True, False]

# Step 3: Apply the filter NumPy picks only the elements matching True (index 0 and 2)
newarr = arr[x]

# Step 4: Display the result shows only [41, 43]
print(newarr)  # Output: [41 43]
```



- **Boolean Indexing / Filtering:** Passing a list of boolean values (`True`/`False`) into array brackets acts as a filter mask.
- **Selection Rule:** Elements aligned with `True` are kept; elements aligned with `False` are discarded.
- **Length Matching:** The boolean list or mask must match the length of the array dimension being indexed to avoid an `IndexError`.

The example above will return [41, 43] because the new array contains only the values where the filter array had the value True, in this case, index 0 and 2.

#### Create a filter array that will return only values higher than 42:

```python
# Step 1: Create the initial array of numbers
arr = np.array([41, 42, 43, 44])

# Step 2: Initialize an empty list to store our True/False filter decisions
filter_arr = []

# Step 3: Loop through each item in the array to test a condition
for element in arr:
    # If the number is strictly greater than 42, keep it (True); otherwise, drop it (False)
    if element > 42:
        filter_arr.append(True)
    else:
        filter_arr.append(False)

# Step 4: Apply the boolean mask to the array to extract only the matching items
newarr = arr[filter_arr]

# Step 5: Show the boolean list: [False, False, True, True]
print(filter_arr)

# Step 6: Show the filtered array: [43, 44]
print(newarr)
```



- **Manual Boolean Filtering:** Demonstrates how a filter mask is built step-by-step using standard Python logic (`for` loop + `if-else`).
- **Condition (`element > 42`):**
  - `41 > 42` is `False`
  - `42 > 42` is `False` (since it is equal, not strictly greater)
  - `43 > 42` is `True`
  - `44 > 42` is `True`
- **NumPy Shortcut Note:** In practice, NumPy can replace this entire loop in a single line using vectorized conditions: `filter_arr = arr > 42`.

#### Alternate method (Creating Filter Directly From Array)

```python
# Step 1: Create the initial array of numbers
arr = np.array([41, 42, 43, 44])

# Step 2: Directly apply the condition (vectorized) — NumPy tests every item at once
filter_arr = arr > 42

# Step 3: Use the boolean mask to extract only the elements that met the condition (True)
newarr = arr[filter_arr]

# Step 4: Show the boolean array: [False False  True  True]
print(filter_arr)

# Step 5: Show the filtered array: [43 44]
print(newarr)
```

- **Vectorized Operation:** Instead of using a slow `for` loop, NumPy evaluates `arr > 42` directly across all elements simultaneously.
- **Direct Masking:** `filter_arr` automatically becomes a boolean array of `True` and `False` values matching the exact shape of `arr`.
- **Clean & Fast:** `arr[filter_arr]` extracts only the elements evaluated as `True` (`43` and `44`), representing the standard, high-performance Pythonic way to filter data in NumPy.

#### Create a filter array that will return only even elements from the original array:

```python
# Step 1: Create an array containing numbers 1 through 7
arr = np.array([1, 2, 3, 4, 5, 6, 7])

# Step 2: Initialize an empty list to store the True/False filter decisions
filter_arr = []

# Step 3: Loop through each number in the array to check for even numbers
for element in arr:
    # If the number is evenly divisible by 2 (remainder is 0), mark it True; otherwise, False
    if element % 2 == 0:
        filter_arr.append(True)
    else:
        filter_arr.append(False)

# Step 4: Apply the boolean mask to extract only the matching items (the even numbers)
newarr = arr[filter_arr]

# Step 5: Show the boolean list: [False, True, False, True, False, True, False]
print(filter_arr)

# Step 6: Show the filtered array: [2, 4, 6]
print(newarr)
```



- **Modulo Operator (`%`):** `element % 2 == 0` is the standard check for even numbers because it checks whether dividing by 2 leaves a remainder of 0.
- **Manual Filtering Process:**
  - Odd numbers (`1, 3, 5, 7`) have a remainder of `1`, so they get `False`.
  - Even numbers (`2, 4, 6`) have a remainder of `0`, so they get `True`.
- **NumPy One-Liner Alternative:** Just like the previous example, this whole loop can be written directly using vectorization as `filter_arr = arr % 2 == 0`.

#### Alternate method: (Create a filter array that will return only even elements from the original array)

```python
# Step 1: Create an array containing numbers 1 through 7
arr = np.array([1, 2, 3, 4, 5, 6, 7])

# Step 2: Directly apply the modulo condition across the whole array to find even numbers
filter_arr = arr % 2 == 0

# Step 3: Apply the boolean mask to pull out only the elements evaluated as True
newarr = arr[filter_arr]

# Step 4: Show the boolean array: [False  True False  True False  True False]
print(filter_arr)

# Step 5: Show the filtered array containing only even numbers: [2 4 6]
print(newarr)
```



- **Vectorized Modulo (`arr % 2 == 0`):** NumPy applies the remainder operator `%` to every single element simultaneously without requiring a manual loop.
- **Boolean Output:** Every even number evaluates to `True` (remainder is 0), and every odd number evaluates to `False` (remainder is 1).
- **Direct Indexing:** Passing `filter_arr` into `arr[...]` instantly filters out all elements flagged as `False`, returning only `[2, 4, 6]`.
