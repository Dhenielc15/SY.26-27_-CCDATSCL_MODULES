# Numpy

- NumPy is a Python library.

- it is used for working with arrays.

- NumPy is short for "Numerical Python".

## Why use Numpy?

- In Python we have lists that serve the purpose of arrays, but they are slow to process.

- NumPy aims to provide an array object that is up to 50x faster than traditional Python lists.

- The array object in NumPy is called ndarray, it provides a lot of supporting functions that make working with ndarray very easy.

- Arrays are very frequently used in data science, where speed and resources are very important.

- NumPy arrays are stored at one continuous place in memory unlike lists, so processes can access and manipulate them very efficiently.

- This behavior is called locality of reference in computer science.

- This is the main reason why NumPy is faster than lists. Also it is optimized to work with latest CPU architectures.

```python
# Step 1: Print the array to inspect its structure and elements
print(arr)
# Step 2: Check the data type of the elements stored in the array
print(arr.dtype)
```



- **Inspecting Array Values (`print(arr)`):** Displays the full content and structural arrangement of the array across its dimensions.
- **Data Type Attribute (`.dtype`):** Returns the exact internal type of the stored elements (e.g., `int64`, `int32`, `float64`).
- **Homogeneous Nature of NumPy:** Every element inside a NumPy array must share the exact same data type, allowing NumPy to store data contiguously in memory and execute fast, vectorized operations.
- **Automatic Type Inference:** When creating arrays without explicitly specifying `dtype`, NumPy automatically infers the most appropriate data type based on the input values.

```python
# Step 1: Create a 1D NumPy array with string elements
arr = np.array(['apple', 'banana', 'cherry'])

# Step 2: Print the array elements
print(arr)  
# Output: ['apple' 'banana' 'cherry']

# Step 3: Inspect the automatically inferred data type
print(arr.dtype)  
# Output: <U6
```



- **String Storage in NumPy:** NumPy supports string arrays, but unlike standard Python lists that hold references to dynamically sized strings, NumPy allocates fixed-width slots for strings to maintain contiguous memory blocks.
- **Decoding `<U6`:**
  - `<` indicates **little-endian** byte ordering.
  - `U` stands for **Unicode string**.
  - `6` represents the **maximum character length** among all elements in the array (in this case, `"banana"` and `"cherry"` both have 6 characters).
- **Fixed-Length Truncation Trap:** Because the dtype is locked at `<U6`, assigning a longer string later (e.g., `arr[0] = 'watermelon'`) will silently truncate the value to the first 6 letters (`'waterm'`).
