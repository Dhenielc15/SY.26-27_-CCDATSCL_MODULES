# Joining Numpy Arrays

Joining means putting contents of two or more arrays in a single array.

## Using concatenate()

#### 1-D Array

```python
import numpy as np

# Step 1: Create two separate 1D NumPy arrays
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

# Step 2: Join the two arrays end-to-end into a single array using np.concatenate()
# Note: Arrays must be passed inside a tuple: (arr1, arr2)
arr = np.concatenate((arr1, arr2))

# Step 3: Print the combined array: [1 2 3 4 5 6]
print(arr)
```



- **Array Joining (`np.concatenate`):** Combines a sequence of existing arrays along an existing axis into one single array.
- **Tuple Requirement:** The arrays to be joined must be passed inside a tuple or list (e.g., `(arr1, arr2)`), otherwise NumPy will treat the second array as the axis parameter and raise an error.
- **Default Axis (Axis 0):** For 1D arrays, concatenation simply appends the elements of `arr2` to the end of `arr1`, resulting in `[1, 2, 3, 4, 5, 6]`.
- **Memory Allocation:** Concatenation creates a brand new array in memory rather than modifying the original arrays in-place.

#### 2-D Array

```python
import numpy as np

# Step 1: Create two 2D arrays, each with shape (2, 2)
arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])

# Step 2: Join the arrays horizontally (column-wise) using axis=1
# Arrays must be enclosed in a tuple: (arr1, arr2)
arr = np.concatenate((arr1, arr2), axis=1)

# Step 3: Print the concatenated 2D array
print(arr)
# Output:
# [[1 2 5 6]
#  [3 4 7 8]]
```



- **Horizontal Concatenation (`axis=1`):** In 2D arrays, `axis=0` refers to rows (vertical stacking), while `axis=1` refers to columns (horizontal stacking). Setting `axis=1` appends columns side by side.
- **Visual Merging:**
  - Row 0 joins `[1, 2]` + `[5, 6]` $\rightarrow$ `[1, 2, 5, 6]`
  - Row 1 joins `[3, 4]` + `[7, 8]` $\rightarrow$ `[3, 4, 7, 8]`
- **Shape Transformation:** 
  - Inputs: Both arrays have shape `(2, 2)`.
  - Output: The resulting array has shape `(2, 4)` because the number of columns increased while the number of rows remained the same.
- **Dimension Matching Rule:** To concatenate along `axis=1`, the arrays must have the exact same number of rows (`axis=0`).

## Using stack()

```python
import numpy as np

# Step 1: Create two 1D NumPy arrays
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

# Step 2: Stack arrays along a new second axis (axis=1), turning 1D arrays into columns
# Arrays must be passed inside a tuple: (arr1, arr2)
arr = np.stack((arr1, arr2), axis=1)

# Step 3: Print the resulting 2D array
print(arr)
# Output:
# [[1 4]
#  [2 5]
#  [3 6]]
```



- **New Dimension Creation (`np.stack`):** Unlike `np.concatenate` which joins arrays along an existing axis, `np.stack` joins them along a **brand new axis**, increasing the array's dimensionality from 1D to 2D.
- **Column Stacking via `axis=1`:** 
  - `axis=0` would stack them as rows: `[[1, 2, 3], [4, 5, 6]]` (shape `(2, 3)`).
  - `axis=1` stacks corresponding pairs side-by-side as columns: `[1, 4]`, `[2, 5]`, `[3, 6]` (shape `(3, 2)`).
- **Shape Evolution:**
  - Inputs: Two 1D arrays with shape `(3,)`.
  - Output: One 2D array with shape `(3, 2)`.
- **Requirement:** All input arrays passed to `np.stack()` must have the exact same shape.

## Stacking along rows

```python
import numpy as np

# Step 1: Create two 1D NumPy arrays
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

# Step 2: Stack arrays horizontally (along rows / end-to-end) using np.hstack()
# Arrays must be passed inside a tuple: (arr1, arr2)
arr = np.hstack((arr1, arr2))

# Step 3: Print the combined 1D array: [1 2 3 4 5 6]
print(arr)
```



- **Horizontal Stacking (`np.hstack`):** Stacks arrays horizontally, meaning it connects them sequence-by-sequence along the horizontal direction.
- **Behavior on 1D Arrays:** For 1D arrays, `np.hstack()` simply concatenates them end-to-end along the first axis, producing the same result as `np.concatenate((arr1, arr2))`.
- **Output Shape:**
  - Inputs: Two 1D arrays of shape `(3,)`.
  - Output: A single extended 1D array `[1 2 3 4 5 6]` with shape `(6,)`.
- **Contrast with `np.stack`:** Unlike `np.stack()` which creates a new 2D dimension, `np.hstack()` preserves the existing 1D dimension.

## Stacking along columns

```python
# Step 1: Create two 1D NumPy arrays
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

# Step 2: Stack arrays vertically (row-wise / on top of each other) using np.vstack()
# Arrays must be passed inside a tuple: (arr1, arr2)
arr = np.vstack((arr1, arr2))

# Step 3: Print the resulting 2D array
print(arr)
# Output:
# [[1 2 3]
#  [4 5 6]]
```



- **Vertical Stacking (`np.vstack`):** Stacks arrays vertically (row by row / on top of each other), effectively turning multiple 1D arrays into distinct rows of a new 2D matrix.
- **Dimensionality Upgrade:** Unlike `np.hstack()` which keeps 1D arrays flat, `np.vstack()` promotes 1D arrays into 2D by treating each input array as an individual row.
- **Shape Evolution:**
  - Inputs: Two 1D arrays with shape `(3,)`.
  - Output: A 2D array with shape `(2, 3)` (2 rows, 3 columns).
- **Dimension Matching Rule:** All input arrays must have the same length along their columns so that the stacked rows align properly.

## Stacking Along Height (depth)

```python
# Step 1: Create two 1D NumPy arrays
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

# Step 2: Stack arrays depth-wise (along the 3rd axis / depth) using np.dstack()
# Arrays must be passed inside a tuple: (arr1, arr2)
arr = np.dstack((arr1, arr2))

# Step 3: Print the resulting 3D array
print(arr)
# Output:
# [[[1 4]
#   [2 5]
#   [3 6]]]
```



- **Depth-Wise Stacking (`np.dstack`):** Stacks arrays along the third dimension (depth / `axis=2`), simulating stacking layers or sheets of data one behind the other.
- **Automatic 3D Promotion:** For 1D and 2D arrays, `np.dstack()` automatically reshapes them into 3D before combining:
  - A 1D array of shape `(N,)` is first treated as shape `(1, N, 1)`.
  - Stacking two such arrays produces shape `(1, N, 2)`.
- **Shape Evolution:**
  - Inputs: Two 1D arrays with shape `(3,)`.
  - Output: A 3D array with shape `(1, 3, 2)` (1 block, 3 rows, 2 depth-wise columns).
- **Practical Use Case:** Frequently used in image processing to stack separate color channels (e.g., combining Red, Green, and Blue 2D layers into a single RGB 3D array).
