# Copy and View

## Copy


```python
import numpy as np

# Step 1: Create the original array with numbers 1 to 5
arr1 = np.array([1, 2, 3, 4, 5])

# Step 2: Make an independent copy so arr2 gets its own space in memory
arr2 = arr1.copy()

# Step 3: Change the first item of arr1 to 42
arr1[0] = 42

# Step 4: Show arr1 — the first element changes to 42
print(arr1)  # Output: [42  2  3  4  5]

# Step 5: Show arr2 — it stays the same because .copy() protects it
print(arr2)  # Output: [1 2 3 4 5]
```



- **Independent Copy (`arr1.copy()`)**: Creates a completely separate duplicate of the array in a new memory location.
- **Isolated Mutation (`arr1[0] = 42`)**: Changing an element in `arr1` only affects `arr1` because `arr2` does not share the same data buffer.
- **Data Protection**: `arr2` remains untouched `[1, 2, 3, 4, 5]`, proving that `.copy()` safely prevents unwanted side effects when modifying arrays.

## View

The view does not own the data and any changes made to the view will affect the original array, and any changes made to the original array will affect the view.

```python
# Step 1: Create the original array with numbers 1 to 5
arr1 = np.array([1, 2, 3, 4, 5])

# Step 2: Create a VIEW — this does NOT copy the data, both arrays share the same memory
arr2 = arr1.view()

# Step 3: Change the first item of arr1 to 42
arr1[0] = 42

# Step 4: Show arr1 — updated to [42, 2, 3, 4, 5]
print(arr1)  # Output: [42  2  3  4  5]

# Step 5: Show arr2 — ALSO changed to [42, 2, 3, 4, 5] because it looks at the exact same data
print(arr2)  # Output: [42  2  3  4  5]
```



- **Shared Memory (`arr1.view()`):** Creates a new array object that looks at the exact same data buffer without copying memory.
- **Synchronized Mutation (`arr1[0] = 42`):** Changing the original array immediately modifies the view as well, because both reference the same memory address.
- **Memory Efficiency vs. Risk:** `.view()` saves memory by not duplicating data, but modifying one array alters both.
