# Iterating Numpy Arrays

Iterating means going through elements one by one.

#### 1-D Array

````python
import numpy as np

# Step 1: Create a basic 1D NumPy array with three integers
arr = np.array([1, 2, 3])

# Step 2: Iterate directly over the array elements using a standard for loop
for x in arr:
    # Print each element one by one on a new line
    print(x)
````


- **Direct Iteration:** Iterating over a 1D NumPy array behaves just like iterating through a standard Python list, moving sequentially through each element from index `0` to the end.
- **Scalar Extraction:** In each iteration, variable `x` temporarily holds a single scalar value (`1`, then `2`, then `3`).
- **Sequential Output:** Each call to `print(x)` displays the current item on a separate line, producing:
  ```text
  1
  2
  3

#### 2-D Array

````python
# Step 1: Create a 2D array (matrix) with 2 rows and 3 columns
arr = np.array([[1, 2, 3], [4, 5, 6]])

# Step 2: Iterate over the array using a single for loop
# In a 2D array, the outer loop steps through entire rows, not individual numbers
for x in arr:
    # Print each row (1D sub-array) one at a time
    print(x)
````



- **Row-by-Row Iteration:** A standard `for` loop on a 2D array iterates over the first axis (axis 0), meaning each pass yields an entire 1D sub-array (a row), not individual scalar values.
- **Iteration Breakdown:**
  - 1st loop: `x` holds `[1, 2, 3]` (Row 1).
  - 2nd loop: `x` holds `[4, 5, 6]` (Row 2).
- **Accessing Individual Elements:** To reach every single number, you would either need a nested loop (`for y in x: print(y)`) or NumPy's built-in iterator `np.nditer(arr)`.

````python
import numpy as np

# Step 1: Create a 2D array (matrix) with 2 rows and 3 columns
arr = np.array([[1, 2, 3], [4, 5, 6]])

# Step 2: Outer loop iterates over each row (1D sub-array)
for x in arr:
    # Step 3: Inner loop iterates over each individual element inside the current row
    for y in x:
        # Print each scalar value one by one on a separate line
        print(y)
````



- **Nested Loops for Multidimensional Traversal:** Since a 2D array consists of rows and columns, traversing down to individual scalar elements requires two nested loops.
- **Loop Breakdown:**
  - **Outer Loop (`for x in arr`):** Extracts each row as a 1D sub-array (`[1, 2, 3]`, then `[4, 5, 6]`).
  - **Inner Loop (`for y in x`):** Unpacks each number within that row sequentially (`1, 2, 3` followed by `4, 5, 6`).
- **Alternative in NumPy:** While nested loops work, they can become slow and tedious for higher dimensions ($3D$, $4D$). NumPy provides `np.nditer(arr)` to traverse every scalar element in any array using just a single loop.
