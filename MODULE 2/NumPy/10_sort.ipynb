# Sorting Arrays

Sorting means putting elements in an ordered sequence.

Ordered sequence is any sequence that has an order corresponding to elements, like numeric or alphabetical, ascending or descending.

#### Integers

```python
import numpy as np

# Step 1: Create a 1D NumPy array with unsorted integers
arr = np.array([3, 2, 0, 1])

# Step 2: Sort the array in ascending order
# np.sort() returns a new sorted copy without modifying the original array
sorted_arr = np.sort(arr)

# Step 3: Print the sorted array
print(sorted_arr)  
# Output: [0 1 2 3]
```



- **1D Array Sorting (`np.sort`):** Rearranges the elements of the array into ascending numerical order by default.
- **Copy vs. In-Place:**
  - `np.sort(arr)` generates and returns a **new sorted copy** of the array, leaving the original `arr` untouched.
  - To sort directly in-place without allocating a new array, use the ndarray method `arr.sort()`.
- **Default Algorithm:** Under the hood, NumPy typically uses the **QuickSort** algorithm ($\mathcal{O}(N \log N)$ average time complexity), with options to specify `kind='mergesort'` (stable) or `kind='heapsort'`.
- **Descending Order Technique:** NumPy does not offer a direct `reverse=True` flag for `np.sort()`. To sort in descending order, slice with a negative step: `np.sort(arr)[::-1]`.

#### String

```python
# Step 1: Create a 1D NumPy array of string elements
arr = np.array(['banana', 'cherry', 'apple'])

# Step 2: Sort the string elements in lexicographical (alphabetical) order
# np.sort() returns a new sorted copy of the array
sorted_arr = np.sort(arr)

# Step 3: Print the sorted array
print(sorted_arr)  
# Output: ['apple' 'banana' 'cherry']
```



- **Lexicographical Sorting:** When applied to text data, `np.sort()` sorts strings in alphabetical (ASCII/Unicode lexicographical) order rather than numerical value.
- **Underlying NumPy String Dtype:**
  - NumPy automatically detects these elements as fixed-width Unicode strings (e.g., `<U6`, representing the maximum string length of 6 characters in `'banana'` and `'cherry'`).
  - Because string lengths are fixed in NumPy arrays, string operations remain vectorized and memory-aligned.
- **Case Sensitivity:** Lexicographical sorting is case-sensitive based on ASCII values (all uppercase letters `A-Z` sort before lowercase letters `a-z`).
- **Non-Destructive Execution:** As with numerical arrays, `np.sort(arr)` leaves the original array `arr` unchanged in memory and returns a brand-new sorted array.

#### Boolean

```python
# Step 1: Create a 1D NumPy array with boolean values
arr = np.array([True, False, True])

# Step 2: Sort the boolean elements in ascending order
# np.sort() treats False as 0 and True as 1
sorted_arr = np.sort(arr)

# Step 3: Print the sorted boolean array
print(sorted_arr)  
# Output: [False  True  True]
```



- **Boolean Type Promotion & Sorting:** NumPy treats boolean values according to standard binary integer mappings: `False` evaluates to `0` and `True` evaluates to `1`.
- **Ascending Sort Order:** Since `0 < 1`, all `False` elements are positioned before all `True` elements, transforming `[True, False, True]` into `[False, True, True]`.
- **Dtype Preservation:** Even though the comparison relies on numeric values ($0$ and $1$), the returned array retains the original `bool` data type (`dtype=bool`), not integer type.
- **Practical Application (Partitioning Flags):** Frequently used in data filtering pipelines to group inactive records (`False`) away from active or flagged records (`True`).

#### 2-D array

```python
# Step 1: Create a 2D array of shape (2, 3) with unsorted values
arr = np.array([[3, 2, 4], [5, 0, 1]])

# Step 2: Sort the array along the default axis (axis=-1, the last axis / rows)
# np.sort() returns a sorted copy without modifying the original array in-place
sorted_arr = np.sort(arr)

# Step 3: Print the sorted 2D array
print(sorted_arr)
# Output:
# [[2 3 4]
#  [0 1 5]]
```



- **Independent Row-Wise Sorting (`np.sort`):** By default, `np.sort()` sorts along the last axis (`axis=-1`), which corresponds to each individual row in a 2D matrix.
- **Row Separation:** Elements are reordered strictly within their respective rows; values never move between different rows:
  - Row 0: `[3, 2, 4]` becomes `[2, 3, 4]`
  - Row 1: `[5, 0, 1]` becomes `[0, 1, 5]`
- **Non-Destructive Return:** Unlike Python's `.sort()` method on lists or `arr.sort()` (which modifies in-place), top-level `np.sort(arr)` returns a **new sorted array copy**, preserving the original `arr`.
- **Controlling the Sort Axis:**
  - `axis=-1` or `axis=1` (default for 2D): Sorts horizontally along columns (each row independently).
  - `axis=0`: Sorts vertically down each column across rows.
  - `axis=None`: Flattens the array first and returns a sorted 1D array (`[0, 1, 2, 3, 4, 5]`).
