# Generate Random Number

#### Generate random int

```python
from numpy import random

# Step 1: Generate a single pseudo-random integer between 0 (inclusive) and 100 (exclusive)
x = random.randint(100)

# Step 2: Print the generated random number
print(x)  # Output: e.g., 42 (varies each run)
```



- **NumPy Random Module (`numpy.random`):** A dedicated sub-package containing optimized tools for random sampling, permutations, and probability distributions.
- **Half-Open Interval $[0, 100)$:** Passing a single integer $N$ (`random.randint(100)`) generates a discrete uniform integer where:
  - Minimum possible value is **0** (inclusive).
  - Maximum possible value is **99** (exclusive of 100).
- **Default Output Shape:** When no `size` parameter is specified, `random.randint()` returns a single scalar integer rather than an array.
- **Specifying Ranges:** To define custom lower and upper boundaries, pass two arguments: `random.randint(low, high)` (e.g., `random.randint(10, 50)` generates values from 10 to 49).

#### Generate random float

```python
# Step 1: Generate a pseudo-random floating-point number between 0.0 (inclusive) and 1.0 (exclusive)
x = random.rand()

# Step 2: Print the generated float value
print(x)  # Output: e.g., 0.472945... (varies each run)
```



- **Uniform Float Sampling (`random.rand()`):** Draws samples from a continuous uniform distribution over the half-open interval $[0.0, 1.0)$.
- **Scalar Return:** Calling `random.rand()` without arguments returns a single Python `float` instead of an array.
- **Dimensional Arguments:** Passing dimensions directly as comma-separated integers generates multidimensional arrays without needing tuples (e.g., `random.rand(3, 2)` produces a $3 \times 2$ array of floats).
- **Difference from `randint`:** While `randint()` produces discrete whole numbers, `rand()` yields continuous decimal values, making it ideal for probabilities, weights, or random scaling factors.

## Random Distribution

Data Distribution is a list of all possible values, and how often each value occurs.

#### Generate a 1-D array containing 100 values, where each value has to be 3, 5, 7 or 9.

The probability for the value to be 3 is set to be 0.1

The probability for the value to be 5 is set to be 0.3

The probability for the value to be 7 is set to be 0.6

The probability for the value to be 9 is set to be 0



#### The choice() method allows us to specify the probability for each value.

```python
from numpy import random

# Step 1: Draw a 1D array of 100 random samples from a given list with defined probabilities
# 3 has a 10% chance (0.1)
# 5 has a 30% chance (0.3)
# 7 has a 60% chance (0.6)
# 9 has a 0% chance (0.0) will never appear
x = random.choice([3, 5, 7, 9], p=[0.1, 0.3, 0.6, 0.0], size=(100))

# Step 2: Print the resulting array of 100 sampled elements
print(x)
```



- **Custom Discrete Sampling (`random.choice`):** Generates random values chosen strictly from an explicit set of options rather than an arbitrary range.
- **Probability Weighting (`p=[...]`):** 
  - Allows assigning individual likelihoods to each option.
  - The sum of all probabilities in `p` must strictly equal `1.0` (100%), otherwise NumPy raises a `ValueError`.
- **Zero-Probability Exclusion:** Elements with a probability of `0.0` (like `9`) are guaranteed never to appear in the generated output.
- **Output Control (`size=(100)`):** Specifies the total number of samples drawn. Setting a tuple like `size=(3, 5)` would produce a 2D matrix of sampled values instead of a 1D array.

#### Same example as above, but return a 2-D array with 3 rows, each containing 5 values.

```python
from numpy import random

# Step 1: Draw random samples from the list using custom probabilities,
# shaped directly into a 2D array of 3 rows and 5 columns
# 3 has a 10% chance (0.1)
# 5 has a 30% chance (0.3)
# 7 has a 60% chance (0.6)
# 9 has a 0% chance (0.0) — never appears
x = random.choice([3, 5, 7, 9], p=[0.1, 0.3, 0.6, 0.0], size=(3, 5))

# Step 2: Print the resulting 3x5 matrix
print(x)
# Example Output:
# [[7 7 5 7 5]
#  [7 5 7 7 3]
#  [5 7 7 7 7]]
```



- **Multidimensional Sampling (`size=(3, 5)`):** Passing a shape tuple into the `size` parameter structures the sampled outputs directly into a 2D matrix (3 rows $\times$ 5 columns) in a single operation.
- **Independent Draws:** Each cell inside the $3 \times 5$ grid is an independent random draw governed by the weights defined in `p`.
- **Distribution Pattern:** With a 60% probability assigned to `7`, it will visually dominate the matrix, while `5` appears moderately often, `3` rarely appears, and `9` will never be present.
- **Grid Synthesis:** Demonstrates how NumPy eliminates nested loops or manual reshaping when populating synthetic matrices according to explicit probability distributions.

## Random Permutations
A permutation refers to an arrangement of elements. e.g. [3, 2, 1] is a permutation of [1, 2, 3] and vice-versa.

#### Randomly shuffle elements of following array:

The shuffle() method makes changes to the original array.

```python
import numpy as np
from numpy import random

# Step 1: Create an ordered 1D NumPy array
arr = np.array([1, 2, 3, 4, 5])

# Step 2: Randomly shuffle the elements IN-PLACE
# Note: The original array 'arr' itself is permanently modified
random.shuffle(arr)

# Step 3: Print the modified array
print(arr)  # Output: e.g., [3 1 5 2 4] (order varies each run)
```



- **In-Place Mutation (`random.shuffle`):** `random.shuffle()` alters the sequence of elements directly inside the original array buffer; it returns `None` rather than producing a new copy.
- **Memory Footprint:** Because it mutates the existing memory buffer in-place, it is highly memory-efficient when working with massive datasets.
- **Multidimensional Behavior:** When applied to 2D or higher-dimensional arrays, `shuffle()` only shuffles along the first axis (it rearranges entire rows, not individual elements within rows).
- **Alternative Function (`permutation`):** If you need to keep the original array untouched, use `random.permutation(arr)`, which returns a shuffled duplicate while leaving the source array intact.

#### Generate a random permutation of elements of following array:

The permutation() method returns a re-arranged array (and leaves the original array un-changed).

```python
# Step 1: Create an ordered 1D NumPy array
arr = np.array([1, 2, 3, 4, 5])

# Step 2: Generate a shuffled copy of the array using random.permutation()
# This returns a newly rearranged array without altering the original
print(random.permutation(arr))  # Output: e.g., [4 1 5 2 3] (varies each run)

# Step 3: Print the original array to verify it remains completely unchanged
print(arr)  # Output: [1 2 3 4 5]
```



- **Non-Destructive Shuffling (`random.permutation`):** Unlike `random.shuffle()`, which modifies data in-place, `random.permutation()` creates and returns a brand-new reordered array copy.
- **Original Data Immutability (`print(arr)`):** The source array `arr` retains its original order (`[1, 2, 3, 4, 5]`), making this safe to use when the initial dataset must be preserved for subsequent operations.
- **Integer Argument Feature:** Passing an integer $N$ directly instead of an array (e.g., `random.permutation(5)`) generates a randomly permuted sequence of integers from `0` to $N-1$ (`np.arange(N)` shuffled).
- **Core Choice (`shuffle` vs. `permutation`):**
  - Use `random.shuffle()` for memory savings when in-place modification is intended.
  - Use `random.permutation()` when downstream analysis depends on the unmodified original array.
