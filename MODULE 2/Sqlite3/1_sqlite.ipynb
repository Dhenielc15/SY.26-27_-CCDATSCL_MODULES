# SQLite
SQLite is a software library that implements a self-contained, serverless, zero-configuration, transactional SQL database engine.

### Keywords
- **SQLite**: A lightweight, serverless SQL database engine that stores the whole database in a single file. It needs no setup, which makes it great for learning and small applications.
- **`sqlite3`**: Python's built-in module for working with SQLite databases. No installation is needed.
- **`sqlite3.connect()`**: Opens a connection to a database file. If the file does not exist, a new database is created.
- **`CREATE TABLE`**: A SQL command that creates a new table with named columns and data types (e.g., `INT`, `TEXT`, `REAL`).
- **`PRIMARY KEY` / `NOT NULL`**: `PRIMARY KEY` marks the column that uniquely identifies each row. `NOT NULL` means the column cannot be left empty.
- **CRUD**: Stands for Create, Read, Update, Delete, the four basic database operations. In SQL these are `INSERT`, `SELECT`, `UPDATE`, and `DELETE`.
- **`conn.execute()`**: Runs a SQL statement on the database.
- **Cursor**: The object returned by a `SELECT` query. You loop through it to read each row, and each row is a tuple you access by position (`row[0]`, `row[1]`, ...).
- **`conn.commit()`**: Saves changes (insert, update, delete) permanently to the database. Without it, the changes can be lost.
- **`conn.total_changes`**: The total number of rows inserted, updated, or deleted since the connection was opened.
- **`conn.close()`**: Closes the database connection when you are done.

```python
import sqlite3  # Import Python's built-in sqlite3 module for working with SQLite databases

conn = sqlite3.connect('ali.db')  # Connect to the database file 'ali.db' (it is created if it does not exist)
print("Opened database successfully")  # Print a message to confirm the connection
```

## Create a table

```python
conn.execute('''CREATE TABLE COMPANY
         (ID INT PRIMARY KEY     NOT NULL,  -- ID: whole number, unique identifier of each row, cannot be empty
         NAME           TEXT    NOT NULL,  -- NAME: text, cannot be empty
         AGE            INT     NOT NULL,  -- AGE: whole number, cannot be empty
         ADDRESS        CHAR(50),  -- ADDRESS: text up to 50 characters, can be empty
         SALARY         REAL);''')  # Run a CREATE TABLE statement to make a table named COMPANY with 5 columns
print("Table created successfully")  # Print a message to confirm the table was created
```

## Insert record into a table

```python
# Insert row 1 (Rona) into COMPANY; the backslash continues the SQL string on the next line
conn.execute("INSERT INTO COMPANY (ID,NAME,AGE,ADDRESS,SALARY) \
      VALUES (1, 'Rona', 31, 'California', 20000.00 )")

# Insert row 2 (Sly) into COMPANY
conn.execute("INSERT INTO COMPANY (ID,NAME,AGE,ADDRESS,SALARY) \
      VALUES (2, 'Sly', 25, 'Texas', 15000.00 )")

# Insert row 3 (Mariz) into COMPANY
conn.execute("INSERT INTO COMPANY (ID,NAME,AGE,ADDRESS,SALARY) \
      VALUES (3, 'Mariz', 23, 'Norway', 20000.00 )")

# Insert row 4 (Jaymie) into COMPANY
conn.execute("INSERT INTO COMPANY (ID,NAME,AGE,ADDRESS,SALARY) \
      VALUES (4, 'Jaymie', 25, 'Rich-Mond ', 65000.00 )");

conn.commit()  # Save (commit) all the inserted rows permanently to the database
print("Records created successfully")  # Print a message to confirm the records were inserted
```

## Select records from a table

```python
cursor = conn.execute("SELECT id, name, address, salary , age from COMPANY")  # Run a SELECT query; the result is a cursor we can loop through

for row in cursor:  # Loop through each row returned by the query
   print("ID = ", row[0])  # Print the 1st selected column (id)
   print("NAME = ", row[1])  # Print the 2nd selected column (name)
   print("ADDRESS = ", row[2])  # Print the 3rd selected column (address)
   print("SALARY = ", row[3])  # Print the 4th selected column (salary)
   print("AGE = ", row[4])  # Print the 5th selected column (age)
```

## Update a record in a table

```python
conn.execute("UPDATE COMPANY set SALARY = 25000.00 where ID = 1")  # Change the salary of the employee with ID 1 to 25000
conn.commit()  # Save the change permanently to the database
print("Total number of rows updated :", conn.total_changes)  # Show the total number of rows changed since the connection was opened

cursor = conn.execute("SELECT id, name, address, salary from COMPANY")  # Select the records again to see the update
for row in cursor:  # Loop through each returned row
   print ("ID = ", row[0])  # Print the id
   print ("NAME = ", row[1])  # Print the name
   print ("ADDRESS = ", row[2])  # Print the address
   print ("SALARY = ", row[3], "\n")  # Print the salary followed by a blank line

print ("Operation done successfully")  # Print a message to confirm the operation finished
```

## Delete operation

```python
conn.execute("DELETE from COMPANY where ID = 2;")  # Delete the employee whose ID is 2
conn.commit()  # Save the deletion permanently to the database
print ("Total number of rows deleted :", conn.total_changes)  # Show the total number of rows changed since the connection was opened

cursor = conn.execute("SELECT id, name, address, salary from COMPANY")  # Select the records again to see the remaining rows
for row in cursor:  # Loop through each returned row
   print ("ID = ", row[0])  # Print the id
   print ("NAME = ", row[1])  # Print the name
   print ("ADDRESS = ", row[2])  # Print the address
   print ("SALARY = ", row[3], "\n")  # Print the salary followed by a blank line

print ("Operation done successfully")  # Print a message to confirm the operation finished
```

```python
conn.close()  # Close the database connection when we are done
```
